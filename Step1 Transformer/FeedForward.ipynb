```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

前馈神经网络FFN

```python
class FFN(nn.Module):
    def __init__(self,dim,hidden_dim,dropout):
        super().__init__()

        #第一个FC，从输入层到隐藏层
        self.w1 = nn.Linear(dim,hidden_dim,bias=False)

         #第二个FC，从隐藏层到输入层
        self.w2 = nn.Linear(hidden_dim,dim,bias=False)

        #Dropout 防止过拟合
        self.dropout = nn.Dropout(dropout)

    def forward(self,x):
        #输入先后通过FC1,Relu，FC2,dropout
        return self.dropout(self.w2(F.relu(self.w1(x))))
```

```python
# 超参数
dim = 512
hidden_dim = 512 * 4
dropout = 0.1
seq_len = 128
batch_size = 8

ffn =FFN(dim,hidden_dim,dropout)

#构造输入
x = torch.randn(batch_size,seq_len,dim)
x_out = ffn(x)
print(x.shape)
print(x_out.shape)
print(x_out)
```
