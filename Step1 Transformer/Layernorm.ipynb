层归一化（LayerNorm)

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class LayerNorm(nn.Module):
    def __init__(self,dim,eps=1e-6):
        super().__init__()

        '''
        层归一化，用于对一个句子内每一个token的归一化操作
        [x - mean(x) / (std(x) + eps)] * gamma + beta
        '''

        #定义gamma和beta
        self.gamma = nn.Parameter(torch.ones(dim))
        self.beta = nn.Parameter(torch.zeros(dim))
        self.eps = eps

    def forward(self,x):
        #x.shape =[batch_size,max_len,dim]
        #计算均值和方差
        mean= x.mean(-1,keepdim=True)#mean:[batch_size,max_len,1]
        std = x.std(-1, keepdim=True) # mean: [batch, max_len, 1]

        return self.gamma * (x - mean) / (std + self.eps) + self.beta
```

```python
# 超参数
batch_size = 32
seq_len = 10
dim = 256

# 构造输入
features = torch.randn(batch_size, seq_len, dim)

# 实例化layer norm
ln = LayerNorm(dim)
features_output = ln(features)

print("输入的shape: ", features.shape)
print("输出的shape: ", features_output.shape)

print("输入的均值: ", features.mean(-1)[0])
print("输出的均值: ", features_output.mean(-1)[0])
```

```python
print("输入的方差: ", features.std(-1)[0])
print("输出的方差: ", features_output.std(-1)[0])
```
