```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```python
#BPE分词

text = "hello,大模型！"

#编码
#1.把文本变成tokens
tokens = tokenizer.tokenize(text)
print("tokens:",tokens)

#2.把tokens转成id
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print("token id：",token_ids)

#解码
#3.把id转换成tokens
tokens = tokenizer.convert_ids_to_tokens(token_ids)
print("tokens:",tokens)

#4.把tokens拼接成文本
decoded_text = tokenizer.convert_tokens_to_string(tokens)
print("decoded_text",decoded_text)
```

```python
for token in tokens:
    byte_sequence = list(token.encode("utf-8"))
    byte_str = bytes(byte_sequence).decode("utf-8",errors="replace")
    print(f"Token: '{token}' -> Bytes: '{byte_sequence}' -> Decoded: '{byte_str}'")
```

WordPiece分词器

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "hello,大模型！"

#编码
#1.把文本变成tokens
tokens = tokenizer.tokenize(text)
print("tokens:",tokens)

#2.把tokens转成id
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print("token id：",token_ids)

#解码
#3.把id转换成tokens
tokens = tokenizer.convert_ids_to_tokens(token_ids)
print("tokens:",tokens)

#4.把tokens拼接成文本
decoded_text = tokenizer.convert_tokens_to_string(tokens)
print("decoded_text",decoded_text)
```

encode和decode方法

```python
tokenizer = AutoTokenizer.from_pretrained("gpt2")
text = "hello,大模型！"

token_ids = tokenizer.encode(text)
print("token_ids", token_ids)

decoded_text = tokenizer.decode(token_ids)
print("decoded_text",decoded_text)
```
