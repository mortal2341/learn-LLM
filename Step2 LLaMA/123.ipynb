```python
! pip install tiktoken
import tiktoken
import torch
import json
from pathlib import Path
from tiktoken.load import load_tiktoken_bpe
! pip install modelscope
# 加载分词器
tokenizer_model_path = "LLM-Research/Meta-Llama-3-8B-Instruct/original/tokenizer.model"

# LLaMA3的分词器的特殊字符表
special_tokens = [
                     "<|begin_of_text|>",
                     "<|end_of_text|>",
                     "<|reserved_special_token_0|>",
                     "<|reserved_special_token_1|>",
                     "<|reserved_special_token_2|>",
                     "<|reserved_special_token_3|>",
                     "<|start_header_id|>",
                     "<|end_header_id|>",
                     "<|reserved_special_token_4|>",
                     "<|eot_id|>",  # end of turn
                 ] + [f"<|reserved_special_token_{i}|>" for i in range(5, 256 - 5)]

# 加载 mergeranks
merge_ranks = load_tiktoken_bpe(tokenizer_model_path)

# 正则串定义BPE算法的切分规则
# 例如，pat_str 可能需要设计在空格处拆分文本，或将某些标点符号保留在单词后，或分离出特定符号，数字
tokenizer = tiktoken.Encoding(
    name=Path(tokenizer_model_path).name,
    pat_str=r"(?i:'s|'t|'re|'ve|'m|'ll|'d)|[^\r\n\p{L}\p{N}]?\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]+[\r\n]*|\s*[\r\n]+|\s+(?!\S)|\s+",
    mergeable_ranks=merge_ranks,
    special_tokens={token: len(merge_ranks) + i for i, token in enumerate(special_tokens)},
)
len(merge_ranks)
import random

random.sample(list(merge_ranks.keys()), 10)
tokenizer.n_vocab
tokenizer.encode("你好!")
tokenizer.decode(tokenizer.encode("你好!"))
## transformers库来加载
!pip install transformers
from transformers import AutoTokenizer

tokenizer_2 = AutoTokenizer.from_pretrained("LLM-Research/Meta-Llama-3-8B-Instruct/")
print(tokenizer_2.encode("你好!"))
print(tokenizer_2.decode(tokenizer.encode("你好!")))
tokenizer_2.decode([128000])
```
