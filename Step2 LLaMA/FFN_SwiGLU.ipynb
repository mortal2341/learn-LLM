```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

FFN_SwiGLU 手写实现

```python
class FFNSwiGLU(nn.Module):
    def __init__(self,input_dim,hidden_dim):
        super(FFNSwiGLU, self).__init__()

        #2 * input_dim * ffn_dim = 3 * input_dim * swi_hidden_dim
        #swi_hidden_dim = (2 * ffn_dim) / 3
        hidden_dim = int(2 * hidden_dim / 3)

        self.up = nn.Linear(input_dim,hidden_dim)
        self.down = nn.Linear(hidden_dim,input_dim)
        self.gate = nn.Linear(input_dim,hidden_dim)

    def forward(self,x):
        return self.down(self.up(x) * F.silu(self.gate(x)))
```

```python
input_dim = 1024
hidden_dim = 4096
ffn_swiglu = FFNSwiGLU(input_dim, hidden_dim)

# 构造输入
x = torch.randn(1, input_dim)
output = ffn_swiglu(x)
print("input shape: ", x.shape)
print("output shape: ", output.shape)
print(output)
```
